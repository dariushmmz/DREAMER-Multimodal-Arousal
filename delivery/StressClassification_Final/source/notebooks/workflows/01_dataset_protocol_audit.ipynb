# Corrected dataset and leakage audit
This notebook audits the tracked DREAMER three-class **arousal** index. It does not claim that arousal is a stress diagnosis.

```python
from pathlib import Path
import sys, json
import pandas as pd

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / 'pyproject.toml').is_file(): ROOT = ROOT.parent
if not (ROOT / 'pyproject.toml').is_file(): raise RuntimeError('Repository root not found')
sys.path.insert(0, str(ROOT / 'src'))
from stress_classification.config import load_config
from stress_classification.data import validate_window_index
from stress_classification.splits import SubjectSplit

config = load_config(ROOT / 'configs/dreamer_arousal_3class.yaml')
index = pd.read_csv(ROOT / 'data/processed/window_index.csv')
split = SubjectSplit.from_mapping(config['split'])
validate_window_index(index)
split.validate(index.subject_id.unique())
print(json.dumps(json.loads((ROOT / 'data/processed/dataset_summary.json').read_text()), indent=2))
```
